```python
import pandas as pd
import unicodedata
```

```python
df = pd.read_excel("CONSOLIDADO_PROYEC.xlsx")
df.info()
df.head()
```

```python
def quitar_acentos(texto):
    return ''.join(
        caracter for caracter in unicodedata.normalize('NFD', texto)
        if unicodedata.category(caracter) != 'Mn'
    )
```

```python

df["Nombre_del_proyecto_de_investigación"] = (
    df["Nombre_del_proyecto_de_investigación"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.lower()
    .apply(quitar_acentos)
)
```

```python
df["Resumen_del_objetivo_del_proyecto_de_investigación"] = (
    df["Resumen_del_objetivo_del_proyecto_de_investigación"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.lower()
    .apply(quitar_acentos)
)
```

```python

patron = (
    #Para buscar inteligencia artificial
    r"inteligencia artificial"
    r"|artificial intelligence"
    r"|\bia\b"
    r"|i\.a\.?"

    #Bucar super cómputo
    r"|super ?computo"
    r"|supercomputacion"
    r" |high performance computing"
    r"|supercomputer"

    #Buscar satelite
    r"|satelites?"
    r"|satelitale?s?"
    r"|satellites?"

    #Buscar machine learning
     r"|machine ?learning"
     r"|aprendizajes? computacionale?s?"
     r"|aprendizaje de maquinas?"
     r"|aprendizaje maquinas?"
     r"|aprendizaje automatico"
     r"|aprendizaje mecanico"

     #Deep learning
    r"|deep ?learning"
    r"|aprendizaje profundo"

    #Redes neuronales
    r"|redes? neuronale?s?"
    r"|neural network"

    #IA generativa
    r"|modelos? generativos?"
    r"|generative models?"

    #LLM y modelos fundacionales
    r"|(llms)"
    r"|(llm)"
    r"|llms?"
    r"|modelos? de lenguaje"
    r"|modelos? de lenguajes? de gran tamaño"
    r"|modelos? de lenguaje de gran escala"
    r"|modelos? de lenguajes? extensos?"
    r"|large language models?"
    r"|foundation models?"
    r"|foundation model"

    #lenguaje
    r"|procesamiento de lenguaje natural"
    r"|natural language processing"
    r"|comprension del lenguaje natural"

    #Otras formas en las que puede aparecer según el archivo en word

    r"|modelos? de aprendizaje"
    r"|modelos? de descripción"
    r"|modelos? secuenciale?s? de aprendizaje"
    r"|inteligencias? computacionale?s?"
    r"|vision computacional"
    r"|generacion de modelos?"
    r"|self ?learning"
    r"|aprendizaje federado"
    r"|rede?s? generativas?"
    r"|rede?s? generativas? adversarias?"
    r"|modelos? de prediccion"
    r"|modelos? computacionale?s?"
    r"|computo de alto desempeño"
    r"|realidad ?virtual"
    r"|algoritmos? de aprendizajes?"
    r"|predictore?s? ?estructurale?s?"
    r"|agentes? ?conversacionale?s?"
    r"|diagnosticos? automaticos?"
    r"|diagnoticos? automatizado?"
    r"|convolucione?s? deformables?"
    )
```

```python
filtro = (
    df["Nombre_del_proyecto_de_investigación"].str.contains(
        patron, case=False, na=False, regex=True
    )
    |
    df["Resumen_del_objetivo_del_proyecto_de_investigación"].str.contains(
        patron, case=False, na=False, regex=True
    )
)
```
