```python
import re

# Los textos ya están sin acentos ni "ñ"; los patrones deben estarlo también
# (por ejemplo "descripción" -> "descripcion", "desempeño" -> "desempeno")
patrones = {nombre: quitar_acentos(p) for nombre, p in patrones.items()}


def palabras_encontradas(texto, regex):
    """Devuelve la lista de palabras/frases del texto que coinciden con el patrón."""
    return [m.group(0) for m in regex.finditer(texto)]


# Buscar cada patrón y generar un Excel por patrón
for capitulo, patron in patrones.items():
    regex = re.compile(patron, flags=re.IGNORECASE)

    # Palabras que se encontraron en cada fila
        # Palabras encontradas en cada columna
    por_columna = {
        col: texto_busqueda[col].apply(lambda t: palabras_encontradas(t, regex))
        for col in columnas_busqueda
    }

    # Todas las palabras de la fila (sumando ambas columnas)
    encontradas = pd.Series(
        [sum(listas, []) for listas in zip(*por_columna.values())],
        index=df.index,
    )
    filtro = encontradas.str.len() > 0

    if not filtro.any():
        print(f"{capitulo}: 0 registros encontrados (no se genera archivo)")
        print("-----------------------------------")
        continue

    # Hoja 1: filas encontradas + la palabra del patrón que coincidió
    resultado = df.loc[filtro].copy()
    resultado.insert(0, "Fila_en_Excel_original", resultado.index + 2)
    resultado.insert(
        1,
        "Palabra_encontrada",
        encontradas[filtro].apply(lambda lista: ", ".join(sorted(set(lista)))),
    )
        # Opcional: en qué columna se encontró cada palabra
    for i, col in enumerate(columnas_busqueda):
        resultado.insert(
            2 + i,
            f"Encontrada_en_{col}",
            por_columna[col][filtro].apply(lambda l: ", ".join(sorted(set(l)))),
        )
    resultado = resultado.sort_values("Palabra_encontrada")

    # Hoja 2: resumen (cuántos registros tuvo cada palabra)
    resumen = (
        encontradas[filtro]
        .apply(lambda lista: sorted(set(lista)))
        .explode()
        .value_counts()
        .rename_axis("Palabra_encontrada")
        .reset_index(name="Num_registros")
    )

    # Exportar un Excel por patrón
    ruta_archivo = carpeta_resultados / f"{capitulo}.xlsx"
    with pd.ExcelWriter(ruta_archivo, engine="openpyxl") as writer:
        resultado.to_excel(writer, sheet_name="Resultados", index=False)
        resumen.to_excel(writer, sheet_name="Resumen", index=False)

    print(f"{capitulo}: {len(resultado)} registros encontrados")
    print(f"Archivo generado: {ruta_archivo}")
    print("-----------------------------------")
```
